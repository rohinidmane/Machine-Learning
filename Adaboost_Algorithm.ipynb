# Adaboost Algorithm

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
# Load Iris dataset
iris = load_iris()
X = iris.data
y = iris.target

# For binary classification, let's use only 2 classes (e.g., Setosa and Versicolor)
X = X[y != 2]
y = y[y != 2]

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Use Decision Tree as base estimator
base_estimator = DecisionTreeClassifier(max_depth=1)

# Initialize AdaBoost classifier
ada_clf = AdaBoostClassifier(estimator=base_estimator, n_estimators=50, learning_rate=1.0, random_state=42)

# Fit the model
ada_clf.fit(X_train, y_train)
```

```python
# Predict on test data
y_pred = ada_clf.predict(X_test)

# Accuracy
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# Load the full Iris dataset (all 3 classes)
iris = load_iris()
X = iris.data
y = iris.target

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Initialize base learner and AdaBoost
base_model = DecisionTreeClassifier(max_depth=1)
ada_clf = AdaBoostClassifier(estimator=base_model, n_estimators=50, learning_rate=1.0, random_state=42)

# Train AdaBoost
ada_clf.fit(X_train, y_train)

# Predictions and evaluation
y_pred = ada_clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred, target_names=iris.target_names))
```

```python
# Predict a new flower based on user-provided values

# Format:
# [sepal length, sepal width, petal length, petal width]

new_flower = [[5.1, 3.5, 1.4, 0.2]]

# Make prediction
prediction = ada_clf.predict(new_flower)

# Convert predicted number to flower name
print("Predicted Class:", prediction[0])
print("Predicted Flower:", iris.target_names[prediction[0]])
```
