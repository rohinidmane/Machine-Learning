# Multicollinearity

```python
import pandas as pd
import numpy as np
from statsmodels.stats.outliers_influence import variance_inflation_factor

# --- Step 1: Create a dataset ---
np.random.seed(10)
X1 = np.random.rand(100)
X2 = 2 * X1 + np.random.rand(100) * 0.1   # Highly correlated with X1
X3 = np.random.rand(100)                   # Independent feature

df = pd.DataFrame({
    'X1': X1,
    'X2': X2,
    'X3': X3
})

# --- Step 2: Correlation matrix ---
print("Correlation Matrix:")
print(df.corr(), "\n")

# --- Step 3: Calculate VIF for each variable ---
vif_data = pd.DataFrame()
vif_data["Feature"] = df.columns
vif_data["VIF"] = [variance_inflation_factor(df.values, i) for i in range(df.shape[1])]

print("Variance Inflation Factor (VIF):")
print(vif_data)
```

The correlation between X1 and X2 = 0.9988, which is extremely high.

This means:

X1 and X2 contain almost the same information.

X3 is not highly correlated with either X1 or X2.


Common interpretation:

* VIF = 1 → No multicollinearity
* VIF < 5 → Generally acceptable
* VIF 5–10 → High multicollinearity
* VIF > 10 → Severe multicollinearity
